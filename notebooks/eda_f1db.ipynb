```python
import sys
from pathlib import Path

# Get the project root (up one level from 'notebooks')
project_root = Path.cwd().parent

# Add the project root to sys.path if it's not already there
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

# Now you can import from the src directory
from src.database import F1DataManager

# Initialize the manager
db = F1DataManager()
#query = "SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%' ORDER BY name"
query = """
SELECT name, type
FROM sqlite_master
WHERE type IN ('table', 'view')
  AND name NOT LIKE 'sqlite_%'
ORDER BY type, name
"""
tables = db.query(query)
print(tables)
```

```python
# Try to find info for a particular race, Round 1 2025 (Bahrain)
query = "SELECT id, official_name FROM race WHERE year = 2025 AND official_name LIKE '%Bahrain%'"
race_info = db.query(query)
print(race_info) 
```

```python
# view available fields within race result
query = "PRAGMA table_info(race_data);"
out = db.query(query)
with open("race_data_names.txt", 'w') as f:
    for _, row in out.iterrows():
        f.write(f"{row['name']}\n")
```
